## **Custom Tools (Method 1)**

#### **@tool decorator**

```python
from langchain_core.tools import tool
```

```python
# Step 1 -- create a function

def multiply(a,b):
  """Multiply two numbers"""
  return a*b
```

```python
# Step 2 -- adding type hints

def multiply(a: int,b: int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
# Step 3 -- adding tool decorotor
@tool
def multiply(a: int,b: int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
def multiply1(a,b):
  """Multiply two numbers"""
  return a*b
```

```python
result = multiply.invoke({"a": 5, "b": 3})
print(result)
```

```python
multiply(5,3)
```

```python
multiply1(5,4)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
multiply.args_schema.model_json_schema()
```

## **Custom Tools (Method 2)**
#### **StructuredTool & Pydantic**

```python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number")
    b: int = Field(required=True, description="The second number")
```

```python
def multiply_func(a: int, b: int) -> int:
    return a * b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
result = multiply_tool.invoke({'a':3, 'b':5})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

## **Custom Tools (Method 3)**
#### **BaseTool**

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':5})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

## Toolkit

```python
from langchain_core.tools import tool
```

```python
# Custom tool
@tool
def add(a: int, b: int) -> int:
  """Add two numbers"""
  return a+b

@tool
def multiply(a: int, b: int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
class MathToolKit:
  def get_tools(self):
    return [add, multiply]
```

```python
toolkit = MathToolKit()
tools = toolkit.get_tools()

for tool in tools:
  print(tool.name, "->", tool.description)
```
