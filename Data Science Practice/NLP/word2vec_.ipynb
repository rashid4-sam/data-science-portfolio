```python
import gensim
from gensim.models import Word2Vec,KeyedVectors
import kagglehub
```

```python
path = kagglehub.dataset_download("leadbest/googlenewsvectorsnegative300/versions/2")
```

```python
import os

# List the contents of the downloaded directory
print(os.listdir(path))
```

```python
import os

model_path = os.path.join(path, 'GoogleNews-vectors-negative300.bin')
model = KeyedVectors.load_word2vec_format(model_path, binary=True, limit=500000)
```

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# king, queen, woman, girl, boy, man, water
words = ['queen', 'woman', 'girl', 'boy', 'man', 'king', 'water']
embeddings = np.array([
    model['queen'][0:10],
    model['woman'][0:10],
    model['girl'][0:10],
    model['boy'][0:10],
    model['man'][0:10],
    model['king'][0:10],
    model['water'][0:10],
                       ])

df = pd.DataFrame(embeddings, index=words)
```

```python
# Create a hashmap
plt.figure(figsize=(18,16))
heatmap = sns.heatmap(df, cmap='crest')
plt.show()
```
