Using the re Library for Special Character Removal

```python
import re
input_str = "Hello. How are you?"
clean_str = re.sub(r'[^a-zA-Z0-9\s]', '', input_str)
print(clean_str)
```

Using Spacy for Special Character Removal

```python
import spacy
nlp = spacy.load('en_core_web_sm')
input_str = "Hello. How are you?"
def clean_text(text):
    doc = nlp(text)
    return ' '.join(token.text for token in doc if token.is_alpha or token.is_space)
clean_str = clean_text(input_str)
print(clean_str)
```

```python
# Using NLTK for Special Character Removal

import nltk
nltk.download('punkt')
nltk.download("punkt_tab")
input_str = "Hello. How are you?"
tokens = nltk.word_tokenize(input_str)
clean_tokens = [token for token in tokens if token.isalnum()]
clean_str = ' '.join(clean_tokens)
print(clean_str)
```
