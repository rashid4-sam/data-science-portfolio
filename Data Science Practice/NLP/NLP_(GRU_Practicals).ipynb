```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
# Initiate the Tokenizer

tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([text])
```

```python
len(tokenizer.word_index)
```

```python
for sentence in text.split('\n'):
  print(sentence)
```

```python
for sentence in text.split('\n'):
  print(tokenizer.texts_to_sequences([sentence])[0])
```

```python
input_sequences = []

for sentence in text.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1, len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])
```

```python
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]
y = padded_input_sequences[:,-1]
```

```python
X
```

```python
y
```

```python
tokenizer.word_index
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y, num_classes=576)
```

```python
y.shape
```

```python
X.shape
```

### **Model Building**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense
```

```python
model = Sequential()
model.add(Embedding(input_dim=576, output_dim=100, input_shape=(266,)))
model.add(GRU(150))
model.add(Dense(576, activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.summary()
```

```python
X.shape
```

```python
y.shape
```

```python
model.fit(X,y,epochs=100)
```

#### **Test the model**

```python
text2 = "Data"

# tokenization
token_text = tokenizer.texts_to_sequences([text2])[0]
# padding
padded_text = pad_sequences([token_text], maxlen=33, padding='pre')
# model prediction
model.predict(padded_text)
```

```python
import numpy as np
pos = np.argmax(model.predict(padded_text))
```

```python
tokenizer.word_index
```

```python
for word, index in tokenizer.word_index.items():
  if index==pos:
    print(word)
```
