```python
# Implementing Contraction Expansion from Scratch
import re

def expand_contractions(text):
    contractions_pattern = {
        "can't": "cannot",
        "won't": "will not",
        "it's": "it is",
        "weren't": "were not",
        "I'm": "I am",
        "couldn't": "could not"
    }
    for contraction, expanded in contractions_pattern.items():
        text = re.sub(contraction, expanded, text)
    return text

text = "I can't believe it's already raining."
expanded_text = expand_contractions(text)
print(expanded_text)
```
